# Data Analysis Demo

This notebook demonstrates basic data analysis:
- Loading data
- Summary statistics
- Visualization

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Sample dataset (you can replace later)
data = {
    "Name": ["A", "B", "C", "D"],
    "Score": [70, 85, 90, 60]
}

df = pd.DataFrame(data)
df
```

```python
df.describe()
```

```python
df.plot(kind='bar', x='Name', y='Score')
plt.title("Scores by Student")
plt.show()
```
